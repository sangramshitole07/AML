## Problem 9: Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84]). Using NumPy boolean indexing, find: Marks greater than 75, Marks between 50 and 80, Number of students who scored below 50, Percentage of students who scored 60 or above. Do not use a Python for loop.

```python
import numpy as np

marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

print("Original marks array:", marks)

# Marks greater than 75
marks_gt_75 = marks[marks > 75]
print("\nMarks greater than 75:", marks_gt_75)

# Marks between 50 and 80 (inclusive of 50, exclusive of 80)
marks_between_50_80 = marks[(marks >= 50) & (marks < 80)]
print("Marks between 50 and 80:", marks_between_50_80)

# Number of students who scored below 50
students_below_50 = np.sum(marks < 50)
print("Number of students who scored below 50:", students_below_50)

# Percentage of students who scored 60 or above
students_60_or_above = np.sum(marks >= 60)
percentage_60_or_above = (students_60_or_above / len(marks)) * 100
print(f"Percentage of students who scored 60 or above: {percentage_60_or_above:.2f}%")
```

## Problem 10: The following array contains marks of four students in three subjects: marks = np.array([[80, 75, 90], [65, 70, 72], [88, 92, 85], [55, 60, 58]]). Calculate: Total marks of each student, Average marks of each student, Average marks in each subject, Student with the highest total, Highest mark in each subject.

```python
import numpy as np

marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])

print("Marks array:\n", marks)

# Total marks of each student (sum across columns, axis=1)
total_marks_per_student = np.sum(marks, axis=1)
print("\nTotal marks of each student:", total_marks_per_student)

# Average marks of each student
average_marks_per_student = np.mean(marks, axis=1)
print("Average marks of each student:", average_marks_per_student)

# Average marks in each subject (sum across rows, axis=0)
average_marks_per_subject = np.mean(marks, axis=0)
print("Average marks in each subject:", average_marks_per_subject)

# Student with the highest total
highest_total_student_index = np.argmax(total_marks_per_student)
print(f"Student with the highest total (index {highest_total_student_index}): {total_marks_per_student[highest_total_student_index]}")

# Highest mark in each subject
highest_mark_per_subject = np.max(marks, axis=0)
print("Highest mark in each subject:", highest_mark_per_subject)
```

## Problem 11: Create: A = np.array([[1, 2], [3, 4]]), B = np.array([[5, 6], [7, 8]]). Perform: Matrix addition, Element-wise multiplication, Matrix multiplication, Transpose of A. Explain the difference between: A * B and A @ B

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)

# Matrix addition
matrix_addition = A + B
print("\nMatrix Addition (A + B):\n", matrix_addition)

# Element-wise multiplication
elementwise_multiplication = A * B
print("\nElement-wise Multiplication (A * B):\n", elementwise_multiplication)

# Matrix multiplication
matrix_multiplication = A @ B
print("\nMatrix Multiplication (A @ B):\n", matrix_multiplication)

# Transpose of A
transpose_A = A.T
print("\nTranspose of A:\n", transpose_A)

print("\nExplanation of A * B vs A @ B:")
print("  `A * B` performs element-wise multiplication (Hadamard product). It multiplies corresponding elements of the two arrays.")
print("  `A @ B` performs matrix multiplication (dot product) according to the rules of linear algebra. For this to work, the number of columns in A must match the number of rows in B.")
```

## Problem 12: Daily temperatures for two weeks are: temperature = np.array([31, 32, 33, 35, 36, 34, 32, 30, 29, 31, 33, 35, 37, 36]). Find: Mean temperature, Maximum and minimum temperatures, Number of days above the average temperature, Temperatures greater than 34°C, Difference between maximum and minimum, Standard deviation.

```python
import numpy as np

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

print("Temperature array:", temperature)

# Mean temperature
mean_temp = np.mean(temperature)
print("\nMean temperature:", mean_temp)

# Maximum and minimum temperatures
max_temp = np.max(temperature)
min_temp = np.min(temperature)
print("Maximum temperature:", max_temp)
print("Minimum temperature:", min_temp)

# Number of days above the average temperature
days_above_average = np.sum(temperature > mean_temp)
print("Number of days above the average temperature:", days_above_average)

# Temperatures greater than 34°C
temps_gt_34 = temperature[temperature > 34]
print("Temperatures greater than 34°C:", temps_gt_34)

# Difference between maximum and minimum
diff_max_min = max_temp - min_temp
print("Difference between maximum and minimum:", diff_max_min)

# Standard deviation
std_dev_temp = np.std(temperature)
print("Standard deviation:", std_dev_temp)
```

## Problem 13: Monthly sales of a company are: sales = np.array([12000, 15000, 17000, 14000, 18000, 21000, 19000, 22000, 25000, 23000, 26000, 28000]). Determine: Total annual sales, Average monthly sales, Month with the highest sales, Month with the lowest sales, Number of months with sales above ₹20,000, Percentage increase from January to December. Hint: Remember that NumPy indexing starts at 0, while month numbering normally starts at 1.

```python
import numpy as np

sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

print("Monthly sales array:", sales)

# Total annual sales
total_annual_sales = np.sum(sales)
print("\nTotal annual sales:", total_annual_sales)

# Average monthly sales
average_monthly_sales = np.mean(sales)
print("Average monthly sales:", average_monthly_sales)

# Month with the highest sales (add 1 for 1-based indexing)
month_highest_sales = np.argmax(sales) + 1
print(f"Month with the highest sales: Month {month_highest_sales} (Sales: {sales[month_highest_sales-1]}) ")

# Month with the lowest sales (add 1 for 1-based indexing)
month_lowest_sales = np.argmin(sales) + 1
print(f"Month with the lowest sales: Month {month_lowest_sales} (Sales: {sales[month_lowest_sales-1]})")

# Number of months with sales above ₹20,000
months_above_20k = np.sum(sales > 20000)
print("Number of months with sales above ₹20,000:", months_above_20k)

# Percentage increase from January to December
january_sales = sales[0]
december_sales = sales[-1]
if january_sales != 0:
    percentage_increase = ((december_sales - january_sales) / january_sales) * 100
    print(f"Percentage increase from January to December: {percentage_increase:.2f}%")
else:
    print("Cannot calculate percentage increase as January sales are zero.")
```

## Problem 21: Take your own example to normalize an array.

```python
import numpy as np

# Example array for normalization (Min-Max Scaling)
# Formula: (X - min(X)) / (max(X) - min(X))

data = np.array([10, 20, 30, 40, 50, 100])
print("Original array:", data)

min_val = np.min(data)
max_val = np.max(data)

# Handle the case where all values are the same to avoid division by zero
if max_val - min_val == 0:
    normalized_data = np.zeros_like(data) # or data / max_val if max_val != 0
else:
    normalized_data = (data - min_val) / (max_val - min_val)

print("Normalized array (Min-Max Scaling):", normalized_data)
```

## Problem 22: Use your own data to standardize an array.

```python
import numpy as np

# Example array for standardization (Z-score Normalization)
# Formula: (X - mean(X)) / std(X)

data = np.array([10, 12, 15, 12, 18, 20, 13, 16, 14, 17])
print("Original array:", data)

mean_val = np.mean(data)
std_val = np.std(data)

# Handle the case where standard deviation is zero to avoid division by zero
if std_val == 0:
    standardized_data = np.zeros_like(data) # All values would be the mean, so they become 0 after subtracting mean
else:
    standardized_data = (data - mean_val) / std_val

print("Standardized array (Z-score Normalization):", standardized_data)
```

## Problem 24: Compare the time required to multiply 1,000,000 numbers using a Python list and a NumPy array.

```python
import numpy as np
import time

num_elements = 1_000_000

# --- Python List ---
python_list_a = list(range(num_elements))
python_list_b = list(range(num_elements))

start_time = time.time()
python_result = [a * b for a, b in zip(python_list_a, python_list_b)]
end_time = time.time()
python_time = end_time - start_time
print(f"Time taken for Python list multiplication: {python_time:.6f} seconds")

# --- NumPy Array ---
numpy_array_a = np.arange(num_elements)
numpy_array_b = np.arange(num_elements)

start_time = time.time()
numpy_result = numpy_array_a * numpy_array_b
end_time = time.time()
numpy_time = end_time - start_time
print(f"Time taken for NumPy array multiplication: {numpy_time:.6f} seconds")

print(f"\nNumPy is approximately {python_time / numpy_time:.2f} times faster than Python lists for this operation.")
```
