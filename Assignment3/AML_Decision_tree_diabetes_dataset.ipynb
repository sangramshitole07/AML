```python
import pandas as pd
import numpy as np
```

```python
# Step 2: Import the dataset
# This method allows uploading a file directly to the Colab environment.
from google.colab import files
uploaded=files.upload()

# Read the uploaded CSV file into a pandas DataFrame
dataset=pd.read_csv("diabetes.csv")

# Display the loaded dataset
dataset
```

```python
#step 3 : detailed info
print("Information about dataset",dataset.info())
print("Decription about dataset ",dataset.describe())
```

```python
print(dataset.columns)
```

```python
#Handle values ,check if there is any duplicate value or null values
dataset.isnull().sum()
```

```python
X= dataset.iloc[:,:-1].values
X
#X=dataset.iloc[:,:-1] shows table like struture like it includes the name of the column and row numbers and when we use value then its converted into array using numpy
```

```python
y=dataset.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
X_test
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier=DecisionTreeClassifier(criterion='entropy',random_state=0)
classifier.fit(X_train,y_train)
```

```python
y_pred=classifier.predict(X_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import confusion_matrix
cm =confusion_matrix(y_test,y_pred)
print('Confusion Matrix')
print(cm)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
print('Accuracy Score',accuracy_score(y_test,y_pred))
print('Precision Score',precision_score(y_test,y_pred))
```

```python
num_classes = cm.shape[0]
labels = np.unique(y_test)

print('\nMetrics per class (One-vs-Rest):')
# Iterate through each class to calculate its individual True Positives, False Positives, etc.
# This uses a 'one-vs-rest' approach, treating one class as positive and all others as negative.
for i, class_label in enumerate(labels):
    # True Positives (TP): Correctly predicted instances of the current class_label
    TP = cm[i, i]

    # False Positives (FP): Instances incorrectly predicted as the current class_label (actually another class)
    FP = cm[:, i].sum() - TP

    # False Negatives (FN): Instances of the current class_label incorrectly predicted as another class
    FN = cm[i, :].sum() - TP

    # True Negatives (TN): Instances correctly predicted as *not* the current class_label
    TN = cm.sum() - (TP + FP + FN)

    print(f'\nClass: {class_label}')
    print(f'  TP: {TP}')
    print(f'  FP: {FP}')
    print(f'  FN: {FN}')
    print(f'  TN: {TN}')
```

```python
new_daibetes = np.array([[6, 148, 72, 35, 0, 33.6, 0.627, 50]]) # Example: a single sample with 8 features
scaled_new_daibetes_sample=sc.transform(new_daibetes)
prediction=classifier.predict(scaled_new_daibetes_sample)
print(prediction)

if prediction==0:
  print('The person is not diabetic')
else:
  print('The person is diabetic')
```
