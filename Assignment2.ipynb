### Question 7: Find students scoring above 80, students scoring between 75 and 90, students belonging to CSE, and CSE students scoring above 80.

```python
# 7. Filtering the DataFrame

print("Students scoring above 80:")
display(df[df['Marks'] > 80])

print("\nStudents scoring between 75 and 90:")
display(df[(df['Marks'] >= 75) & (df['Marks'] <= 90)])

print("\nStudents belonging to CSE:")
display(df[df['Department'] == 'CSE'])

print("\nCSE students scoring above 80:")
display(df[(df['Department'] == 'CSE') & (df['Marks'] > 80)])
```

### Question 8: Display students belonging to either CSE or ECE.

### Question 9: Add a new column called 'Result'. Students scoring 75 or above should receive "Pass", otherwise "Fail".

```python
# 9. Add a new column 'Result'
df['Result'] = np.where(df['Marks'] >= 75, 'Pass', 'Fail')

print("DataFrame with 'Result' column:")
display(df)
```

### Question 10: Assign grades according to the given criteria: ≥ 90 (A), ≥ 80 (B), ≥ 70 (C), < 70 (D).

```python
# 10. Assign grades based on Marks
def assign_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    else:
        return 'D'

df['Grade'] = df['Marks'].apply(assign_grade)

print("DataFrame with 'Grade' column:")
display(df)
```

### Question 12: Consider the given DataFrame `students` with missing values. Find missing values, the number of missing values in each column, and rows containing missing values.

```python
# 12. Handling missing values
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}
students = pd.DataFrame(data)

print("Original DataFrame with missing values:")
display(students)

print("\nMissing values (boolean DataFrame):")
display(students.isnull())

print("\nNumber of missing values in each column:")
display(students.isnull().sum())

print("\nRows containing missing values:")
display(students[students.isnull().any(axis=1)])
```

### Question 13: Replace missing 'Age' with the average Age and missing 'Marks' with the average Marks in the `students` DataFrame.

```python
# 13. Replace missing values with averages

# Calculate average age and marks
avg_age = students['Age'].mean()
avg_marks = students['Marks'].mean()

# Fill missing values
students['Age'].fillna(avg_age, inplace=True)
students['Marks'].fillna(avg_marks, inplace=True)

print("DataFrame after replacing missing values with averages:")
display(students)
```

### Question 14: Remove rows containing missing values from the original `students_original` DataFrame (before filling). This requires recreating `students` for demonstration.

### Question 15: Consider the given `df2` DataFrame with duplicate names. Find duplicates, display duplicate rows, and remove duplicates.

### Question 17: Consider the given DataFrame `df`. Find the average marks by department, maximum marks by department, number of students in each department, and for each department, calculate Mean, Minimum, Maximum, and Count.

```python
# 17. Grouping and aggregation
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun", "Divya"],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE"],
    "Marks": [85, 78, 92, 74, 88, 80]
}
df_group = pd.DataFrame(data)

print("Original DataFrame:")
display(df_group)

print("\nAverage marks by department:")
display(df_group.groupby('Department')['Marks'].mean())

print("\nMaximum marks by department:")
display(df_group.groupby('Department')['Marks'].max())

print("\nNumber of students in each department:")
display(df_group.groupby('Department').size())

print("\nFor each department, Mean, Minimum, Maximum, Count:")
display(df_group.groupby('Department')['Marks'].agg(['mean', 'min', 'max', 'count']))
```

### Question 18: Create two DataFrames, `students_df` and `marks_df`, and combine them using `StudentID`.

### Question 19: Use the provided dataset to perform various analyses: display first five records, find dimensions, average final mark, highest/lowest scoring student, students scoring above 80, average marks by department, students with attendance above 80%, sort by FinalMarks, add 'Result' column, add 'Performance' category, and find the department with the highest average FinalMarks.

```python
# 19. Comprehensive Data Analysis
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df_analysis = pd.DataFrame(data)

print("Original DataFrame:")
display(df_analysis)

print("\nFirst five records:")
display(df_analysis.head(5))

print("\nDimensions of the dataset (rows, columns):")
display(df_analysis.shape)

print("\nAverage final mark:")
display(df_analysis['FinalMarks'].mean())

highest_scorer = df_analysis.loc[df_analysis['FinalMarks'].idxmax()]
print("\nHighest-scoring student:")
display(highest_scorer)

lowest_scorer = df_analysis.loc[df_analysis['FinalMarks'].idxmin()]
print("\nLowest-scoring student:")
display(lowest_scorer)

print("\nStudents scoring above 80:")
display(df_analysis[df_analysis['FinalMarks'] > 80])

print("\nAverage marks by department:")
display(df_analysis.groupby('Department')['FinalMarks'].mean())

print("\nStudents having attendance above 80%:")
display(df_analysis[df_analysis['Attendance'] > 80])

print("\nStudents sorted by FinalMarks:")
display(df_analysis.sort_values(by='FinalMarks', ascending=False))

# Add 'Result' column
df_analysis['Result'] = np.where(df_analysis['FinalMarks'] >= 70, 'Pass', 'Fail')
print("\nDataFrame with 'Result' column:")
display(df_analysis)

# Add 'Performance' category
def assign_performance_category(marks):
    if marks >= 90:
        return 'Excellent'
    elif marks >= 80:
        return 'Very Good'
    elif marks >= 70:
        return 'Good'
    else:
        return 'Needs Improvement'

df_analysis['Performance'] = df_analysis['FinalMarks'].apply(assign_performance_category)
print("\nDataFrame with 'Performance' category:")
display(df_analysis)

# Find the department having the highest average FinalMarks
avg_final_marks_by_dept = df_analysis.groupby('Department')['FinalMarks'].mean()
highest_avg_dept = avg_final_marks_by_dept.idxmax()
print(f"\nDepartment with the highest average FinalMarks: {highest_avg_dept}")
display(avg_final_marks_by_dept)
```

### Question 20: Create an `employees` DataFrame and determine various metrics: average, max/min salary, employee with highest salary, employees earning above ₹60,000, average salary/experience by department, number of employees per department, sort by salary, find IT employees with >3 years experience, and add a 'SalaryLevel' column.

```python
# 20. Employee Data Analysis
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

print("Original Employees DataFrame:")
display(employees)

print("\nAverage salary:")
display(employees['Salary'].mean())

print("\nMaximum salary:")
display(employees['Salary'].max())

print("\nMinimum salary:")
display(employees['Salary'].min())

highest_paid_employee = employees.loc[employees['Salary'].idxmax()]
print("\nEmployee receiving the highest salary:")
display(highest_paid_employee)

print("\nEmployees earning above ₹60,000:")
display(employees[employees['Salary'] > 60000])

print("\nAverage salary for each department:")
display(employees.groupby('Department')['Salary'].mean())

print("\nAverage experience by department:")
display(employees.groupby('Department')['Experience'].mean())

print("\nNumber of employees in each department:")
display(employees.groupby('Department').size())

print("\nEmployees sorted by salary:")
display(employees.sort_values(by='Salary', ascending=False))

print("\nIT employees with more than 3 years of experience:")
display(employees[(employees['Department'] == 'IT') & (employees['Experience'] > 3)])

# Add 'SalaryLevel' column
def assign_salary_level(salary):
    if salary >= 75000:
        return 'High'
    elif 50000 <= salary < 75000:
        return 'Medium'
    else:
        return 'Low'

employees['SalaryLevel'] = employees['Salary'].apply(assign_salary_level)
print("\nDataFrame with 'SalaryLevel' column:")
display(employees)
```
